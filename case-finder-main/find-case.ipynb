# Find similar cases and generate clarification questions

This notebook produces two outputs for comparison:

1. **Baseline questions** based only on the new use case.
2. **Enriched questions** informed by the full text of relevant training cases.

Retrieval uses the summary prompt and embedding model shared with the indexing notebook.

````python
from pathlib import Path
import json
import os
import re

import numpy as np
import pandas as pd
from dotenv import load_dotenv
from openai import OpenAI

from prompts import (
    SUMMARY_SYSTEM_PROMPT,
    SUMMARY_USER_TEMPLATE,
)

load_dotenv()

LLM_MODEL = os.getenv("OPENAI_LLM_MODEL", "gpt-4.1-mini")
EMBEDDING_MODEL = os.getenv("OPENAI_EMBEDDING_MODEL", "text-embedding-3-small")

if not os.getenv("OPENAI_API_KEY"):
    raise RuntimeError("OPENAI_API_KEY is missing. Copy .env.example to .env and add your key.")

client = OpenAI()

from prompts import (
    ENRICHED_QUESTION_SYSTEM_PROMPT,
    ENRICHED_QUESTION_USER_TEMPLATE,
    QUESTION_SYSTEM_PROMPT,
    QUESTION_USER_TEMPLATE,
    QUESTION_LIST_EVALUATION_SYSTEM_PROMPT,
    QUESTION_LIST_EVALUATION_USER_TEMPLATE,
    RELEVANCE_SYSTEM_PROMPT,
    RELEVANCE_USER_TEMPLATE,
)
````

## Input

Replace the example with the new use-case text.

````python
USE_CASE_TEXT = """
We want to create a SaaS that converts custom text input to relevant SQL query
""".strip()

if not USE_CASE_TEXT:
    raise ValueError("USE_CASE_TEXT must not be empty")
````

## Helpers

````python
def summarize_case(case_text: str) -> str:
    response = client.responses.create(
        model=LLM_MODEL,
        instructions=SUMMARY_SYSTEM_PROMPT,
        input=SUMMARY_USER_TEMPLATE.format(case_text=case_text),
    )
    return " ".join(response.output_text.strip().splitlines())


def embed_text(text: str) -> np.ndarray:
    response = client.embeddings.create(model=EMBEDDING_MODEL, input=text)
    return np.asarray(response.data[0].embedding, dtype=float)


def parse_json_array(raw_text: str) -> list:
    """Parse an array while tolerating an accidental Markdown code fence."""
    text = raw_text.strip()
    fenced = re.fullmatch(r"```(?:json)?\s*(.*?)\s*```", text, flags=re.DOTALL | re.IGNORECASE)
    if fenced:
        text = fenced.group(1)
    value = json.loads(text)
    if not isinstance(value, list):
        raise ValueError("The model response must be a JSON array")
    return value


def ask_for_json_array(system_prompt: str, user_prompt: str) -> list:
    response = client.responses.create(
        model=LLM_MODEL,
        instructions=system_prompt,
        input=user_prompt,
    )
    return parse_json_array(response.output_text)


def cosine_similarity(query: np.ndarray, matrix: np.ndarray) -> np.ndarray:
    query_norm = np.linalg.norm(query)
    row_norms = np.linalg.norm(matrix, axis=1)
    if query_norm == 0 or np.any(row_norms == 0):
        raise ValueError("Cannot compute cosine similarity for a zero vector")
    return (matrix @ query) / (row_norms * query_norm)
````

## 1. Baseline questions from the input only

````python
baseline_questions = ask_for_json_array(
    QUESTION_SYSTEM_PROMPT,
    QUESTION_USER_TEMPLATE.format(case_text=USE_CASE_TEXT),
)

print("Baseline questions")
for number, question in enumerate(baseline_questions, start=1):
    print(f"{number}. {question}")
````

## 2. Summarize and retrieve the five nearest cases

````python
INDEX_PATH = Path("data/train/train_index.csv")
TRAIN_DIR = Path("data/train/files")
TOP_K = int(os.getenv("CASE_FINDER_TOP_K", "5"))
if TOP_K < 1:
    raise ValueError("CASE_FINDER_TOP_K must be a positive integer")

if not INDEX_PATH.exists():
    raise FileNotFoundError("Build data/train/train_index.csv with prepare-train-data.ipynb first")

index_df = pd.read_csv(INDEX_PATH)
required_columns = {"id", "summary", "embedding"}
if not required_columns.issubset(index_df.columns):
    raise ValueError(f"Index must contain columns: {sorted(required_columns)}")
if index_df.empty:
    raise ValueError("The training index is empty")

case_summary = summarize_case(USE_CASE_TEXT)
query_embedding = embed_text(case_summary)
embedding_matrix = np.vstack([
    np.asarray(json.loads(value), dtype=float) for value in index_df["embedding"]
])
if embedding_matrix.shape[1] != query_embedding.size:
    raise ValueError(
        "Embedding dimensions differ. Rebuild the index with the configured embedding model."
    )

index_df = index_df.copy()
index_df["similarity"] = cosine_similarity(query_embedding, embedding_matrix)
top_candidates = index_df.nlargest(min(TOP_K, len(index_df)), "similarity").reset_index(drop=True)

print("Input summary:", case_summary)
top_candidates[["id", "similarity", "summary"]]
````

## 3. Let the LLM remove false-positive candidates

Integer positions are used in the prompt so case IDs remain opaque and cannot influence the relevance judgment.

````python
candidate_payload = "\n\n".join(
    f"Candidate {position}: {row.summary}"
    for position, row in enumerate(top_candidates.itertuples(index=False), start=1)
)
selected_positions_raw = ask_for_json_array(
    RELEVANCE_SYSTEM_PROMPT,
    RELEVANCE_USER_TEMPLATE.format(
        target_summary=case_summary,
        candidates=candidate_payload,
    ),
)

valid_positions = set(range(1, len(top_candidates) + 1))
if any(type(value) is not int or value not in valid_positions for value in selected_positions_raw):
    raise ValueError("The relevance response contains an invalid candidate ID")
selected_positions = list(dict.fromkeys(selected_positions_raw))
relevant_candidates = top_candidates.iloc[[position - 1 for position in selected_positions]].copy()

if relevant_candidates.empty:
    print("No candidate passed the LLM relevance filter.")
else:
    display(relevant_candidates[["id", "similarity", "summary"]])
````

## 4. Load complete reference cases and generate enriched questions

````python
case_paths_by_id = {path.stem: path for path in TRAIN_DIR.glob("*.md")}
missing_case_ids = [case_id for case_id in relevant_candidates["id"] if case_id not in case_paths_by_id]
if missing_case_ids:
    raise FileNotFoundError(f"Full case files not found for IDs: {missing_case_ids}")

reference_sections = []
for case_id in relevant_candidates["id"]:
    full_text = case_paths_by_id[case_id].read_text(encoding="utf-8")
    reference_sections.append(f"<reference id={json.dumps(case_id)}>\n{full_text}\n</reference>")

reference_cases = (
    "\n\n".join(reference_sections)
    if reference_sections
    else "No relevant reference cases were found."
)
enriched_questions = ask_for_json_array(
    ENRICHED_QUESTION_SYSTEM_PROMPT,
    ENRICHED_QUESTION_USER_TEMPLATE.format(
        case_text=USE_CASE_TEXT,
        reference_cases=reference_cases,
    ),
)

print("Enriched questions")
for number, question in enumerate(enriched_questions, start=1):
    print(f"{number}. {question}")
````

## Outputs for future evaluation

Use `baseline_questions` and `enriched_questions` as the two systems to compare. `top_candidates` and `relevant_candidates` preserve retrieval diagnostics for later analysis.

````python
results = {
    "input": USE_CASE_TEXT,
    "summary": case_summary,
    "baseline_questions": baseline_questions,
    "retrieved_case_ids": top_candidates["id"].tolist(),
    "relevant_case_ids": relevant_candidates["id"].tolist(),
    "enriched_questions": enriched_questions,
}
results
````

## 5. Evaluate both approaches on the held-out test set

For every Markdown file in `data/test/files`, the following evaluation:

1. generates a baseline list from the test case alone;
2. retrieves and filters similar training cases, then generates an enriched list;
3. randomly assigns the two lists to labels **A** and **B** independently for every case; and
4. asks an LLM acting as the customer to choose the more professional list that better demonstrates understanding of the task.

The evaluator sees the original use case but is not told which generation method produced either list. Ties are allowed. Brief CSV logging records the number of relevant cases found and the winning approach; detailed JSONL logging additionally preserves both question lists, their randomized presentation, and the evaluator's explanation. Logs are written after every case so a partially completed run remains inspectable.

````python
from datetime import datetime, timezone
import csv
import math
import random

TEST_DIR = Path("data/test/files")
EVALUATION_DIR = Path("data/evaluation")
BRIEF_LOG_PATH = EVALUATION_DIR / "question_list_evaluation_brief.csv"
DETAILED_LOG_PATH = EVALUATION_DIR / "question_list_evaluation_detailed.jsonl"
SIGNIFICANCE_LEVEL = 0.05

if not TEST_DIR.exists():
    raise FileNotFoundError(f"Test directory not found: {TEST_DIR}")

test_case_paths = sorted(TEST_DIR.glob("*.md"))
if not test_case_paths:
    raise ValueError(f"No Markdown test cases found in {TEST_DIR}")

EVALUATION_DIR.mkdir(parents=True, exist_ok=True)
RUN_ID = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S.%fZ")
print(f"Run: {RUN_ID}; test cases: {len(test_case_paths)}")
````

````python
def parse_json_object(raw_text: str) -> dict:
    """Parse an object while tolerating an accidental Markdown code fence."""
    text = raw_text.strip()
    fenced = re.fullmatch(r"```(?:json)?\s*(.*?)\s*```", text, flags=re.DOTALL | re.IGNORECASE)
    if fenced:
        text = fenced.group(1)
    value = json.loads(text)
    if not isinstance(value, dict):
        raise ValueError("The model response must be a JSON object")
    return value


def retrieve_relevant_cases(case_text: str) -> tuple[str, pd.DataFrame, pd.DataFrame]:
    summary = summarize_case(case_text)
    query = embed_text(summary)
    if embedding_matrix.shape[1] != query.size:
        raise ValueError(
            "Embedding dimensions differ. Rebuild the index with the configured embedding model."
        )

    scored = index_df.copy()
    scored["similarity"] = cosine_similarity(query, embedding_matrix)
    candidates = scored.nlargest(min(TOP_K, len(scored)), "similarity").reset_index(drop=True)
    payload = "\n\n".join(
        f"Candidate {position}: {row.summary}"
        for position, row in enumerate(candidates.itertuples(index=False), start=1)
    )
    positions = ask_for_json_array(
        RELEVANCE_SYSTEM_PROMPT,
        RELEVANCE_USER_TEMPLATE.format(target_summary=summary, candidates=payload),
    )
    valid_positions = set(range(1, len(candidates) + 1))
    if any(type(value) is not int or value not in valid_positions for value in positions):
        raise ValueError("The relevance response contains an invalid candidate ID")
    positions = list(dict.fromkeys(positions))
    relevant = candidates.iloc[[position - 1 for position in positions]].copy()
    return summary, candidates, relevant


def make_question_lists(case_text: str) -> dict:
    baseline = ask_for_json_array(
        QUESTION_SYSTEM_PROMPT,
        QUESTION_USER_TEMPLATE.format(case_text=case_text),
    )
    summary, candidates, relevant = retrieve_relevant_cases(case_text)

    missing_ids = [case_id for case_id in relevant["id"] if case_id not in case_paths_by_id]
    if missing_ids:
        raise FileNotFoundError(f"Full case files not found for IDs: {missing_ids}")
    sections = [
        f"<reference id={json.dumps(case_id)}>\n"
        f"{case_paths_by_id[case_id].read_text(encoding='utf-8')}\n</reference>"
        for case_id in relevant["id"]
    ]
    references = "\n\n".join(sections) if sections else "No relevant reference cases were found."
    enriched = ask_for_json_array(
        ENRICHED_QUESTION_SYSTEM_PROMPT,
        ENRICHED_QUESTION_USER_TEMPLATE.format(
            case_text=case_text,
            reference_cases=references,
        ),
    )
    return {
        "summary": summary,
        "baseline_questions": baseline,
        "enriched_questions": enriched,
        "retrieved_case_ids": candidates["id"].tolist(),
        "relevant_case_ids": relevant["id"].tolist(),
    }


def numbered_questions(questions: list) -> str:
    return "\n".join(f"{number}. {question}" for number, question in enumerate(questions, start=1))


def evaluate_question_lists(case_text: str, baseline: list, enriched: list) -> dict:
    presented = [
        {"approach": "baseline", "questions": baseline},
        {"approach": "enriched", "questions": enriched},
    ]
    random.SystemRandom().shuffle(presented)
    label_to_approach = {"A": presented[0]["approach"], "B": presented[1]["approach"]}

    response = client.responses.create(
        model=LLM_MODEL,
        instructions=QUESTION_LIST_EVALUATION_SYSTEM_PROMPT,
        input=QUESTION_LIST_EVALUATION_USER_TEMPLATE.format(
            case_text=case_text,
            list_a=numbered_questions(presented[0]["questions"]),
            list_b=numbered_questions(presented[1]["questions"]),
        ),
    )
    judgment = parse_json_object(response.output_text)
    if set(judgment) != {"winner", "reason"}:
        raise ValueError("Evaluation response must contain exactly 'winner' and 'reason'")
    winner_label = judgment["winner"]
    if winner_label not in {"A", "B", "tie"}:
        raise ValueError("Evaluation winner must be 'A', 'B', or 'tie'")
    if not isinstance(judgment["reason"], str) or not judgment["reason"].strip():
        raise ValueError("Evaluation reason must be a non-empty string")

    return {
        "winner_label": winner_label,
        "winner_approach": "tie" if winner_label == "tie" else label_to_approach[winner_label],
        "reason": judgment["reason"].strip(),
        "label_to_approach": label_to_approach,
        "presented_questions": {
            "A": presented[0]["questions"],
            "B": presented[1]["questions"],
        },
    }


def append_evaluation_logs(record: dict) -> None:
    brief_fields = [
        "run_id", "test_case_id", "relevant_case_count", "winner_approach", "winner_label"
    ]
    write_header = not BRIEF_LOG_PATH.exists() or BRIEF_LOG_PATH.stat().st_size == 0
    with BRIEF_LOG_PATH.open("a", encoding="utf-8", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=brief_fields)
        if write_header:
            writer.writeheader()
        writer.writerow({field: record[field] for field in brief_fields})
    with DETAILED_LOG_PATH.open("a", encoding="utf-8") as file:
        file.write(json.dumps(record, ensure_ascii=False) + "\n")


def one_sided_binomial_p_value(successes: int, trials: int, probability: float = 0.5) -> float:
    """Return P(X >= successes) for X ~ Binomial(trials, probability)."""
    if not 0 <= successes <= trials:
        raise ValueError("successes must be between zero and trials")
    return sum(
        math.comb(trials, value)
        * probability**value
        * (1 - probability) ** (trials - value)
        for value in range(successes, trials + 1)
    )
````

````python
evaluation_records = []
for case_number, test_path in enumerate(test_case_paths, start=1):
    test_case_id = test_path.stem
    test_text = test_path.read_text(encoding="utf-8")
    generated = make_question_lists(test_text)
    evaluation = evaluate_question_lists(
        test_text,
        generated["baseline_questions"],
        generated["enriched_questions"],
    )
    record = {
        "run_id": RUN_ID,
        "timestamp_utc": datetime.now(timezone.utc).isoformat(),
        "test_case_id": test_case_id,
        "relevant_case_count": len(generated["relevant_case_ids"]),
        "winner_approach": evaluation["winner_approach"],
        "winner_label": evaluation["winner_label"],
        "label_to_approach": evaluation["label_to_approach"],
        "summary": generated["summary"],
        "retrieved_case_ids": generated["retrieved_case_ids"],
        "relevant_case_ids": generated["relevant_case_ids"],
        "baseline_questions": generated["baseline_questions"],
        "enriched_questions": generated["enriched_questions"],
        "presented_questions": evaluation["presented_questions"],
        "evaluator_reason": evaluation["reason"],
    }
    append_evaluation_logs(record)
    evaluation_records.append(record)
    print(
        f"[{case_number}/{len(test_case_paths)}] {test_case_id}: "
        f"relevant={record['relevant_case_count']}, winner={record['winner_approach']}"
    )
````

### Aggregate result and statistical significance

The primary outcome is the number of test cases won by the enriched approach. Ties are reported but excluded from the significance test. The notebook uses an **exact one-sided binomial test** on non-tied comparisons, with the null hypothesis that baseline and enriched lists are equally likely to win ($p=0.5$), and the alternative that the enriched approach wins more often. A result is called statistically significant when $p < 0.05$.

````python
evaluation_df = pd.DataFrame(evaluation_records)
winner_counts = evaluation_df["winner_approach"].value_counts().reindex(
    ["enriched", "baseline", "tie"], fill_value=0
)
enriched_wins = int(winner_counts["enriched"])
baseline_wins = int(winner_counts["baseline"])
ties = int(winner_counts["tie"])
non_tied = enriched_wins + baseline_wins
p_value = one_sided_binomial_p_value(enriched_wins, non_tied) if non_tied else float("nan")
is_significant = non_tied > 0 and p_value < SIGNIFICANCE_LEVEL

print(f"Test cases: {len(evaluation_df)}")
print(f"Enriched wins: {enriched_wins}")
print(f"Baseline wins: {baseline_wins}")
print(f"Ties: {ties}")
print(f"Exact one-sided binomial p-value: {p_value:.6g}")
print(f"Statistically significant at alpha={SIGNIFICANCE_LEVEL}: {is_significant}")

summary_result = {
    "run_id": RUN_ID,
    "test_case_count": len(evaluation_df),
    "enriched_wins": enriched_wins,
    "baseline_wins": baseline_wins,
    "ties": ties,
    "non_tied_comparisons": non_tied,
    "one_sided_binomial_p_value": p_value,
    "significance_level": SIGNIFICANCE_LEVEL,
    "is_statistically_significant": is_significant,
}
summary_result
````
